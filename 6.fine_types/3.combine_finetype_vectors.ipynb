```r
cts <- list.files('out_rds/EDP1-EDP2-ARB/', pattern = "*/*_\\d+_allcells_finetypesvector", recursive = TRUE, full.names = TRUE)
for (ct in unique(basename(dirname(cts)))) {
    ct_finetypes <- grep(ct, cts, value = TRUE)
    ct_finetypes_vector <- lapply(ct_finetypes, function(f) {readRDS(f)})
    ct_finetypes_vector <- do.call(c, ct_finetypes_vector)
    saveRDS(ct_finetypes_vector, file.path('out_rds/EDP1-EDP2-ARB/', ct, paste0(ct, '_allcells_finetypesvector.rds')))
    message(paste(ct, 'file object saved to', file.path('out_rds/EDP1-EDP2-ARB/', ct, paste0(ct, '_allcells_finetypesvector.rds'))))
}
```
