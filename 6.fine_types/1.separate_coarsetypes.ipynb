# Write each coarse type to its own rds

```r
seu_qc <- readRDS(file = "/data/srlab/AMP_collab/data/early_disease_synovium/xenium/combined/EDP1/allcells_qc_harm_umap_clusters_annotation_tessera.rds")
```

```r
types <- unique(seu_qc$celltypes.coarse)

for (ct in types) {
    cat(ct, "\n"); flush.console()
    cells <- rownames(seu_qc@meta.data)[seu_qc$celltypes.coarse == ct]
    counts <- seu_qc[["RNA"]]$counts[, cells]
    data   <- seu_qc[["RNA"]]$data[, cells]

    seu_sub <- CreateSeuratObject(counts = counts)
    seu_sub[["RNA"]]$data <- data
    seu_sub@meta.data <- seu_qc@meta.data[cells, ]

    cat("\tsaving", length(colnames(seu_sub)), "cells\n"); flush.console()
    ## sanitize filename
    fname <- paste0(
        "out_rds/_",
        ct,
        ".rds"
    )
    saveRDS(seu_sub, fname)
}
```
