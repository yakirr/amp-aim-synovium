Copy files from `level4` directory here to the appropriate places in the local synapse mirror

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os, glob, shutil

basedir = '/data/srlab/AMP_collab/data/early_disease_synovium/'
dest_path = f'{basedir}/xenium'
```

```python
metadata = pd.concat([
    pd.read_csv(f'{basedir}/Xenium_CTRL-SYN-EDP1_assay manifest.csv').assign(subset="CTRL-SYN-EDP1"),
    pd.read_csv(f'{basedir}/Xenium_RA-SYN-EDP1_assay manifest.csv').assign(subset="RA-SYN-EDP1"),
    pd.read_csv(f'{basedir}/Xenium_RA-SYN-EDP2_assay manifest.csv').assign(subset="RA-SYN-EDP2"),
    pd.read_csv(f'{basedir}/Xenium_PsD-SYN-EDP1_assay manifest.csv').assign(subset="PsD-SYN-EDP1"),
    pd.read_csv(f'{basedir}/Xenium_RA-SYN-ARBITRATE_assay manifest.csv').assign(subset="RA-SYN-ARBITRATE")
]).set_index('Sample_ID')
```

```python
for sid in metadata.index:
    s = f'level4/{sid}'
    subset = metadata.loc[sid].subset

    target_dir = f'{dest_path}/Xenium_{subset}/Level_4'
    os.makedirs(target_dir, exist_ok=True)

    h5_file = f'{s}.h5'
    pq_file = f'{s}.parquet'

    h5_exists = os.path.exists(h5_file)
    pq_exists = os.path.exists(pq_file)

    print(subset, s)

    # both missing → red error, continue
    if not h5_exists and not pq_exists:
        print(f"\033[91mERROR: Both files missing for {sid}\033[0m")
        continue

    # one missing → error, stop
    if not h5_exists or not pq_exists:
        missing = h5_file if not h5_exists else pq_file
        print(f"\033[91mERROR: Missing one file for {sid}: {missing}\033[0m")
        raise FileNotFoundError(f"Missing required file for {sid}")

    # both exist → copy
    shutil.copy(h5_file, f'{target_dir}/{sid}.h5')
    shutil.copy(pq_file, f'{target_dir}/{sid}.parquet')
```
