# Write each coarse type to its own rds

```r
seu_qc <- readRDS(file = "/data/srlab/AMP_collab/lakshay-yakir/4.lineages/out_rds/EDP1/allcells_qc_harm_umap_clusters_lineage_annotation.rds")
```

```r
types <- unique(seu_qc$lineage)

for (lineage in types) {
    message(Sys.time(), " | Saving lineage: ", lineage); flush.console()
    cells <- rownames(seu_qc@meta.data)[seu_qc$lineage == lineage]
    counts <- seu_qc[["RNA"]]$counts[, cells]
    data   <- seu_qc[["RNA"]]$data[, cells]

    seu_sub <- CreateSeuratObject(counts = counts)
    seu_sub[["RNA"]]$data <- data
    seu_sub@meta.data <- seu_qc@meta.data[cells, ]

    cat("\tsaving", length(colnames(seu_sub)), "cells\n"); flush.console()
    ## sanitize filename
    dir.create("./out_rds/EDP1/lineages/")
    fname <- paste0(
        "./out_rds/EDP1/lineages/",
        lineage,
        ".rds"
    )
    saveRDS(seu_sub, fname)
}
```
